## Creating select.out for Montserrat VT events

Our goal is to create a subset of the MVOE_ SEISAN database containing only volcano-tectonic events. In this database, those events are identified by volcanic subclass t. SEISAN’s SELECT program can search either CAT files or individual S-files; here we will use the S-file database, which is the more direct choice when working with event metadata and associated waveform references. SEISAN stores event metadata in REA/{DB}/YYYY/MM, one S-file per event, and those S-files include the names of the corresponding waveform files. The waveform files are normally stored in WAV, optionally with a similar year/month structure.  ￼  ￼

We will run SELECT interactively, chose the MVOE database in S-file mode, gave the desired time range, and then set parameter 24 (“Volcanic subclasses”) to t. That will produce select.out, which is a Nordic-format multi-event catalog containing only the selected events, along with index.out and waveform name listings. The SEISAN manual also notes that SELECT outputs Nordic catalogs and that the S-file is the basic per-event unit in the database.  ￼  ￼

This is exactly the right starting point for building a smaller teaching or research subset, as we don't want to change anything in /raid/data/SEISAN_DB. Conceptually, the workflow is:

setup_seisan
select

Then, in SELECT:
	•	choose database: MVOE
	•	choose type: s for S-files
	•	set start and end times
	•	choose parameter 24
	•	enter volcanic subclass: t
	•	press return to execute the search

The result is select.out, a multi-event Nordic file. Each event in that file can then be split back into individual S-files and written into a new database tree such as REA/{dbname}/YYYY/MM. Since the S-files contain waveform references, the corresponding waveform files can then be copied into WAV/{dbname}/YYYY/MM. SEISAN itself has a built-in GET WAV program that can take select.out as input and generate copy commands for the waveform files, but for your FLOVOpy workflow it makes sense to do the same thing directly in Python.  ￼

## Splitting **select.out** back out into a **LOCAL** Seisan database

**subset_select_out.py** is a Python script to split select.out and copy referenced WAV files (this part needs debugging)

This script does the following:

	1.	reads select.out
	2.	splits it into individual Nordic events
	3.	writes each event as an S-file in REA/{dbname}/YYYY/MM
	4.	reads that S-file using your Sfile class
	5.	copies all referenced waveform files into WAV/{dbname}/YYYY/MM
	6.	logs missing waveform files

It assumes:

	- your FLOVOpy package is importable
	- the original SEISAN archive root is known
	- the destination archive root is known
	- you want the new database name padded to 5 characters, as SEISAN expects for on-disk database names. The SEISAN manual notes that database names are 3–5 characters and are padded with _ in the filesystem if shorter.  ￼

```bash
(flovopy_env) glenn@newton:~$ python ~/Developer/CompSciS26/week13/subset_select_out.py \
    --select-out ~/select.out \
    --dest-root "$SEISAN_TOP" \
    --dbname MVOE
```
