
## 📦 SeisComP Data Structure (SDS)

The **SeisComP Data Structure (SDS)** is the global standard directory layout used by observatories to store continuous seismic waveform data in MiniSEED format.

It was designed specifically for:

- Real‑time acquisition systems  
- Long‑term waveform archives  
- Efficient retrieval by time, station, and channel  

Many operational tools (e.g. `slarchive`, `scart`, ObsPy readers, SeedLink systems) are built to work directly with SDS archives. Antelope works seamlessly with it too.
Many digitizers support SDS directly - so when you extract an SD card, the data are already SDs archives.

---

## 🗂️ SDS Directory Layout

SDS stores **one file per channel per day**, organised hierarchically:

```
<SDS_ROOT>/
    YEAR/
        NET/
            STA/
                CHA.TYPE/
                    NET.STA.LOC.CHA.TYPE.YEAR.JDAY
```

Example:

```
/SDS/
  2009/
    AV/
      RDT/
        EHZ.D/
          AV.RDT..EHZ.D.2009.079
```

Where:

| Field | Meaning |
|------|--------|
| `NET` | Network code |
| `STA` | Station |
| `LOC` | Location code |
| `CHA` | Channel |
| `TYPE` | Data type (`D` = waveform) |
| `YEAR` | Year |
| `JDAY` | Julian day |

## 🎯 Why We Store Continuous Data in SDS

I strongly recommend storing *all continuous seismic data* in SDS format because it:

### ✔ Makes data discoverable
Tools can find waveforms instantly by time & station.

### ✔ Enables real-time + archive workflows
The same structure works for:
- live monitoring
- research analysis
- long‑term preservation

### ✔ Prevents chaos
Avoids:

- random filenames  
- mixed sampling rates  
- fragmented day files  
- inconsistent metadata  

### ✔ Is interoperable
ObsPy, SeisComP, EarthScope workflows, and many monitoring systems can read SDS directly.

## 🧠 Conceptually

Instead of this ❌:

```
data/
  station1_day1.mseed
  station2_final.mseed
  copy_of_station1.mseed
```

We use this ✅:

```
SDS/
  2026/AV/RDT/EHZ.D/AV.RDT..EHZ.D.2026.058
```

Which gives:

👉 deterministic structure  
👉 one file per day per channel  
👉 scalable observatory-style archive

# Datasets on our servers

I've have converted many old datasets on our servers from various layouts to SDS archives. The best example of this are data from Montserrat 1995-2008, and 2016-2020.

Digital seismic network Oct 1996 - today:
* Continuous data were captured in 20-minute multi-channel files.
* Triggered event data were also captured in multi-channel files (up to 2 minutes long)
* Captured in Seislog files (until March 2001), and since then Seisan format. 
* Many files are missing - tape backups that failed, or tapes that have degraded.

Analog seismic network Jul 1995 - late 2004.
* Triggered event data from 1995-2001 were captured in SUDS format. These were recorded on Zip disks - many of which developed "click of death".
* Triggered event data from 2001-2004 were captured in Seislog format. Some data lost.
* No continuous data captured.

Issues:
* Giant gaps in the waveform data archives.
* Since Montserrat monitoring began before modern SEED naming conventions were established, raw Montserrat data do not have valid SEED ids.
* Multiple ids for the same station/equipment, no records of what they mean. 
* MVO field technicians did not keep records of equipment installed / dates / gain settings, etc.
* Only a partial record of site coordinates.

Work done at MVO 2002-2003:
* Read all tapes and Zip disks - many unreadable.
* Convert everything to Seisan format.

Work done at USF in 2015:
* Convert each file to MiniSEED
* SEED compliant ids
* Merged into dayfiles, organized as SDS archive
* RESP files converted to Dataless SEED.
* Built a StationXML file.
* All event metadata converted to Seisan S-files.

BUT - not quite finished, and need to do a better job of fixing SEED ids. Actually I need to do a better job of all of this.

And need to go back to the raw tapes and see if more data can be recovered. Steve Malone was able to "bake" tapes from St Helens 1980, and recover continuous data.

Many other datasets converted to SDS:
- Masaya data
- Nevado del Ruiz data
- Kennedy Space Center data
- Alaska Volcano Observatory data (many unrest/eruption sequences)
