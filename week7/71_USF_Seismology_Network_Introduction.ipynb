# Introduction to the USF Seismology Computer Network

## Executive Summary

The USF Seismology Group operates a hybrid computing environment that combines Linux servers, Linux desktops, and macOS laptops to support research, teaching, and real‑time monitoring workflows. The network is designed to balance active research storage, long‑term archival storage, and collaborative access to datasets while maintaining reproducibility and data safety.

Two primary servers form the backbone of the system: **hal9000**, a ZFS‑based active data server used for teaching and current research workflows, and **newton**, a legacy group server that primarily hosts archival datasets and serves as the Ansible configuration management node. Students typically access datasets via read‑only Samba shares from hal9000, while Linux systems share data internally using NFS.

This document provides an overview of the network architecture, server roles, storage strategy, and workflows used in the Computational Seismology course.


## 1. Overview of the Network

The network is designed around several guiding principles:

- Protect scientific data with snapshotting and backups
- Allow easy cross‑platform access to datasets
- Maintain reproducible workflows
- Separate active and archival storage
- Provide a platform for future cluster computing


## 3. Server Roles

### 3.1 hal9000 — Personal Data Server/Workstation

Purchased in 2018, hal9000 is a personal workstation that has evolved into the primary active data server for the lab. It hosts current research datasets, teaching materials, and working directories. The server uses ZFS to provide snapshotting and data integrity.

Key roles:

- Hosts active volcano/rocket datasets
- Provides Samba shares for teaching
- Runs ZFS snapshot and rsync backup workflows to external storage
- Serves as the primary working data hub

### 3.2 newton — Group Primary Server/Workstation

Purchased in 2013 with group funds, newton serves as the archival storage server and configuration management node. It stores legacy datasets and runs Ansible to synchronize user accounts and configurations across Linux systems.

Key roles:

- Ansible control node
- Legacy datasets
- Serve user accounts across Linux network
- Future backup target from Mac desktops
- Legacy datasets
- Secondary backup on data on hal9000


## 4. Data Access Model

Students & Faculty access datasets using **read‑only** Samba shares mounted on their laptops/desktops. This prevents accidental modification while allowing direct access to large datasets without copying.

Linux systems typically access shared storage via NFS, while macOS systems use SMB.


## 5. Account Management

User accounts across Linux machines are synchronized using Ansible, with newton serving as the control node. This ensures consistent usernames, permissions, and environments across the network.
*This is still experimental, and needs further configuration.*


## 6. Future Cluster Computing

The lab plans to build a small Linux cluster using repurposed workstations and surplus hardware. This cluster will support distributed processing of large seismic datasets. Other groups tend to surplus Windows computers when they reach 5 years in age, due to USF IT policy. If they have at least 32 GB RAM, we openly adopt these and install Linux on them, which gives them a new life. We do the same with our Mac computers once they can no longer be updated.

The CIRCE supercomputer at USF is also available for large workloads, though it requires job queueing and is not used regularly by anyone in USF Seismology. This is partly because we cannot mount datasets from newton to CIRCE - all data has to be copied, and disk quotas are relatively small on CIRCE. However, CIRCE is already an excellent choice for data that are stored in the cloud, such as FDSN WS nodes (e.g. Earthscope).

We intend to install our on FDSN WS node on newton, which will eventually enable us to use the ObsPy fdsn client for all workflows, whether we are using datasets archived on newton, or they are coming from Earthscope and other agencies.


## 7. Typical Student Workflow

1. Mount read‑only dataset share from newton (from hal9000 from class)
2. Explore data using Python/ObsPy
3. Store code in GitHub repository
4. Copy small subsets locally, if needed
5. Perform analysis
6. Backup files to newton (not setup yet), or to Box cloud storage, AND to external drive.

**Think about your backup policy!** and also how to share files between everyone on your project/committee.
* GitHub is version-controlled cloud storage for code.
* Box is version-controlled cloud storage for office documents (and small datasets).
* newton is best for backing up value-added/processed data.

Also, if you get new raw datasets, those should probably be uploaded to newton too.
