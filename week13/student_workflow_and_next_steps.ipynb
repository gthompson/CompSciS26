## 10. Example student workflow

A typical workflow for an event analysis session might be:

1. connect to `newton` with NoMachine,
2. open a terminal,
3. run `setup_seisan`,
4. use `select` to define a subset of events,
5. browse that subset with `eev index.out`,
6. open waveforms in `mulplt`,
7. pick phases and amplitudes,
8. save the resulting readings in `REA/{DBNAME}/{YYYY}/{MM}`.

## Suggested workflow after creating a subset catalog

Once you have created a subset catalog such as `select.out`, a sensible next step is to:

1. split the multi-event Nordic file into individual S-files
2. place those S-files in `REA/{DBNAME}/{YYYY}/{MM}`
3. read each S-file to find the referenced waveform file(s)
4. copy the waveform data into `WAV/{DBNAME}/{YYYY}/{MM}`
5. use the resulting subset as a smaller working database for picking, catalog analysis, or machine-learning preparation

This keeps the subset compatible with both SEISAN and Python-based tools.
