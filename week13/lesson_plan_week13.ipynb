🎓 Improved Lesson Plan

Montserrat VT Events: From Waveforms → Catalog

⸻

🧭 Learning Objective

Students will:
	•	understand the end-to-end seismic workflow
	•	compare multiple toolchains (ObsPy, Antelope, SEISAN, Snuffler)
	•	build a consistent event catalog from continuous data

⸻

🧱 Core Structure (concept-first)

Continuous Data
    ↓
Detection
    ↓
Event Definition
    ↓
Picking
    ↓
Location
    ↓
Magnitude
    ↓
Comparison & Validation


⸻

📘 Lesson Modules

⸻

🔹 Module 0 — Dataset & Context (10–15 min)

Goal: Ground everything in a real problem
	•	Introduce Montserrat VT events:
	•	characteristics (impulsive, high-frequency)
	•	typical durations
	•	Show a few example waveforms + spectrograms
	•	Explain station geometry (briefly)

👉 Output:
	•	students know what they’re looking for

⸻

🔹 Module 1 — Detection (ObsPy / flovopy) (30–45 min)

Goal: Understand event detection conceptually

Demonstrate:
	•	STA/LTA theory (brief)
	•	Run detection using your flovopy wrapper

# conceptual example
detections = detect_sta_lta(stream)

Show:
	•	threshold sensitivity
	•	false positives vs missed events
	•	effect of filtering

👉 Output:
	•	list of candidate event times

⸻

🔹 Module 2 — Event Windows (10 min)

Goal: Make detection usable
	•	Extract waveform windows around detections
	•	Explain why this step exists (SEISAN needs it)

Continuous → Event segments


⸻

🔹 Module 3 — Manual Picking (Snuffler FIRST) (45 min)

👉 Important improvement:
Start with Pyrocko Snuffler, not SEISAN.

Why:
	•	modern UI
	•	faster learning curve
	•	immediate visual feedback

Demonstrate:
	•	loading waveforms
	•	picking P and S arrivals
	•	comparing stations

👉 Then briefly show:
	•	SEISAN (eev) as a legacy/operational system

Key teaching point:
	•	picking uncertainty
	•	consistency across stations

⸻

🔹 Module 4 — Location

4A — SEISAN (hyp) (30 min)
	•	Use STATION0.HYP
	•	locate events

4B — Antelope (dbgrassoc) (20 min)
	•	explain travel-time grids
	•	show how association works

👉 Key comparison:

SEISAN	Antelope
1-D model	grid-based
simple	scalable
offline	real-time capable


⸻

🔹 Module 5 — Detection Comparison (Antelope vs ObsPy) (20–30 min)

Now bring in Antelope detection:
	•	run dbdetect
	•	compare with ObsPy results

Discuss:
	•	detection differences
	•	parameter tuning
	•	database vs Python workflows

👉 This is where Antelope shines:
	•	integrated pipeline
	•	reproducibility

⸻

🔹 Module 6 — Magnitude Calculation (20–30 min)

Compare three approaches:

Antelope
	•	dbevproc

SEISAN
	•	hyp / eev

Python / flovopy
	•	your amplitude / energy methods

👉 Key concept:
	•	magnitudes depend on:
	•	amplitude
	•	distance
	•	attenuation assumptions

⸻

🔹 Module 7 — Synthesis & Comparison (15–20 min)

Bring everything together:

Compare:
	•	event times
	•	picks
	•	locations
	•	magnitudes

Ask:
	•	where do results differ?
	•	why?

👉 This is the most important learning moment.

⸻

🔥 Key Improvements Over Your Original Plan

1. Concept-first, not tool-first

Students learn why, not just how

⸻

2. Snuffler before SEISAN
	•	modern → intuitive → confidence boost
	•	SEISAN becomes easier afterward

⸻

3. Separation of detection vs picking

Your original plan mixed them slightly

⸻

4. Antelope positioned correctly
	•	not just another tool
	•	but the integrated system benchmark

⸻

5. Comparison as a final module

Instead of scattered comparisons
