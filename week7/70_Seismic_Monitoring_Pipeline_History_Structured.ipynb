## SAC in Historical Context

SAC (Seismic Analysis Code):

- Developed at Lawrence Livermore National Laboratory
- First released in **1984**
- Written primarily in **C**
- Some numerical routines historically in **Fortran**
- Built for **Unix systems**

### Why SAC Matters

- Predates Perl (1987) and Python (1991)
- Established the classic workflow:
  - Unix
  - SAC
  - Shell scripts
- Still widely used for:
  - Interactive analysis
  - Filtering
  - Visualization
  - Troubleshooting

## Operational Software Emerges (Late 1980s–Early 1990s)

As digital telemetry became common, observatories developed Unix‑based tools.

### Automatic Event Detection

- SEISLOG (used widely in European and volcano observatories)
- MDETECT (USGS)

### Interactive Phase Picking

- SEISAN (University of Bergen)
- SUDSPICK and SUDS utilities (USGS)

### Event Classification

- Manual workflows often performed in SEISAN

## Earthquake Location Programs

These systems typically relied on classic Fortran location codes:

- HYPO71
- HYPOELLIPSE
- HYPOINVERSE
- HYPOCENTER

### Key Feature

SEISAN integrates these programs directly into its workflow.

## SEISAN as an Integrated System

SEISAN was one of the first widely used integrated analysis environments.

### Key Features

- Logical directory structure
- Organized waveform storage
- Instrument response management
- Event metadata organization
- Integrated processing tools

## Emergence of Relational Databases: CSS3.0

The CSS3.0 schema was developed in the late 1980s by the Center for Seismic Studies.

### Why It Was Important

- Introduced relational database concepts to seismology
- Standardized metadata organization
- Separated waveform files from metadata
- Enabled structured querying

### What It Stores

- Arrivals
- Origins
- Stations
- Waveform file references

## Earthworm (Mid‑1990s)

Developed by USGS as a real‑time modular processing system.

### Key Innovations

- Transport rings
- Real‑time message passing
- Modular architecture

### Impact

Gradually replaced earlier detection systems like SEISLOG and MDETECT.

## What Is a Modular Monitoring System?

A modular system divides processing into independent components called modules.

### Modules Can Perform Tasks Such As

- Data acquisition
- Filtering
- Detection
- Picking
- Association
- Location
- Magnitude calculation

### Communication Mechanism

Modules exchange information via:

- Rings (circular buffers)
- Data packets
- Messages

## Rings and Ring Buffers

A ring buffer is a shared memory circular queue.

### Characteristics

- Fixed memory size
- Continuous data flow
- New data overwrites oldest data
- Enables real‑time streaming

## Data Packets and Messages

Each packet is treated as a structured object containing:

- Waveform samples
- Picks
- Detection information
- Processing results

### Advantages

- Loose coupling between modules
- Flexible workflows
- Parallel processing
- Easy system expansion

## Antelope (Mid‑1990s)

Developed by BRTT as a database‑centric monitoring system.

### Built Around

- CSS3.0 relational schema

### Capabilities

- Data ingestion from digitizers
- Event detection
- Automatic picking
- Association
- Event location
- Magnitude calculation
- Bulletin generation

## Analyst Tools in Antelope

Antelope includes powerful analyst review capabilities.

### Example

**dbpick** allows:

- Rapid scanning of continuous waveform data
- Manual pick refinement
- Event identification

## Strength of Antelope

One of its biggest advantages is that:

- Archived data can be processed using the same workflows as real‑time data

## Why Antelope Did Not Become Universal

The main reason:

- It is commercial software

### Result

Many observatories adopted hybrid systems.

## Hybrid Workflows

A common architecture became:

- Earthworm → real‑time detection
- SEISAN → manual processing and catalog refinement

### Example

Used at the Montserrat Volcano Observatory beginning in March 2001.

## Key Takeaways

- Monitoring systems evolved from standalone tools to modular architectures
- SAC remains important for interactive analysis
- SEISAN provided early integrated workflows
- CSS3.0 introduced relational database concepts
- Earthworm introduced modular real‑time processing
- Antelope combined modular processing with databases
- Modern systems build on these foundations
