```python
from obspy.clients.fdsn import Client
from obspy import UTCDateTime

# 1. Initialize the IRIS client
client = Client("IRIS")

# 2. Define network, station, and time range (UTC)
network = "IU"
station = "MAKZ"
# Using BH? to download all three components: BHZ, BHN, BHE
channel = "BH?" 
location = "*"  # Wildcard for location code

# Define start and end times
start = UTCDateTime("2020-06-22T09:25:00")
end = UTCDateTime("2020-06-22T09:30:00")

print(f"Downloading data for {network}.{station} from {start} to {end}...")

# 3. Download the waveform data
try:
    st = client.get_waveforms(network, station, location, channel, start-600, end+600, 
                              attach_response=True)

    print(st)
    st.detrend(type="linear")  # Remove mean
    st.remove_response(output="VEL", water_level=None, pre_filt=(0.5, 0.8, 15, 18))  # Remove instrument response to get velocity
    #st.filter("highpass", freq=1, corners=4, zerophase=True)
    st.trim(start, end)

    # 4. Optional: Plot the data
    st.plot()
    
    # 5. Optional: Save the data to a miniseed file
    # st.write("MAKZ_20200622.mseed", format="MSEED")
    
except Exception as e:
    print(f"An error occurred: {e}")


inv = client.get_stations(network=network, station=station, location=location, channel=channel,
                          starttime=start, endtime=end, level="response")
print(inv[0][0][0])    
```
