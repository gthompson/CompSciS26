# Seismic Monitoring Pipeline Narrative: From Instruments to Modular Systems and Modern Standards

We began this course by looking at the history of computation and seismic instrumentation, and then started to trace the real-time monitoring pipeline from seismometers and digitizers through to analysis and interpretation. Because ObsPy provides a modern and flexible platform for illustrating many monitoring concepts, we first did a crash course on Python and ObsPy fundamentals. Jochen then introduced SAC, which remains a powerful interactive tool for filtering, inspecting, and visualizing waveform data.

SAC fits naturally into the historical perspective because it was developed at Lawrence Livermore National Laboratory, written primarily in C (with some numerical routines historically implemented in Fortran), and first released in 1984 — before scripting languages like Perl (1987) and Python (1991) were available. As a result, the classic workflow of Unix + SAC + shell scripts became deeply embedded in seismology and remains common today. Although SAC has rarely been used as the core real-time monitoring system at observatories, it continues to be widely used for interactive analysis, troubleshooting, and quality control.

In the late 1980s and early 1990s, observatories began developing their own Unix-based operational software as digital telemetry became widespread. Several key tools emerged during this period:

- **Automatic event detection** systems such as **SEISLOG** (used widely in European and volcano observatories) and **MDETECT** (developed at USGS).  
- **Interactive phase picking** tools such as **SEISAN** (developed at the University of Bergen) and **SUDSPICK** and related **SUDS utilities** developed at USGS.  
- **Manual event classification** workflows, often implemented within **SEISAN**.  

These systems were typically used together with Fortran-based earthquake location programs such as **HYPO71**, **HYPOELLIPSE**, **HYPOINVERSE**, and **HYPOCENTER**. SEISAN integrates these location codes so they can be run directly from within the system, making it one of the first widely used integrated earthquake analysis environments. It also introduced a logical directory structure for organizing waveform data, instrument responses, and event metadata.

At the same time, another major development was taking place: the emergence of relational database models for seismic monitoring. The **CSS3.0 schema**, developed in the late 1980s by the **Center for Seismic Studies** as part of nuclear monitoring programs, defined a standardized relational structure for seismic metadata, including arrivals, origins, stations, and waveform file references. CSS3.0 represented a major conceptual shift from file-based workflows to database-centered processing, and it later became the foundation for the Antelope database.

In the mid-1990s, **Earthworm**, developed by USGS, emerged as a modular real-time processing system and gradually replaced earlier detection software such as SEISLOG and MDETECT in many networks. Earthworm introduced the concept of **transport rings** — shared memory circular buffers that allow independent processes to exchange real-time data — and established a highly flexible architecture for real-time monitoring.

Around the same time, **Antelope**, developed by BRTT, became the first widely adopted **database-centric commercial monitoring system** built around the **CSS3.0 schema**. Antelope integrates real-time processing with relational database management, allowing waveform metadata, picks, origins, and magnitudes to be stored and queried consistently.

Both Earthworm and Antelope are examples of **modular systems**, which represent a major architectural evolution in seismic monitoring. In a modular system, processing is divided into independent programs called **modules**. Each module performs a specific task — such as data acquisition, filtering, detection, picking, association, or magnitude calculation — and communicates with other modules by passing **data packets** or **messages** through shared memory **rings** (also called **ring buffers**). These rings act like circular message queues, where new data continuously overwrites older data while maintaining a fixed memory footprint.

Each data packet is treated as an object containing structured information, such as waveform samples, picks, or detection messages. Modules can subscribe to specific packet types, read data from the ring, process it, and then write new messages back to the ring. Because modules are loosely coupled, they can be added, removed, or replaced without redesigning the entire system. This architecture allows highly customized and flexible workflows, supports real-time processing, and makes it possible to run multiple parallel processing pipelines simultaneously.

Antelope builds on this modular architecture while adding a relational database layer based on CSS3.0, enabling seamless integration between real-time processing and long-term archival analysis. It includes modules to ingest data directly from digitizers, detect events, automatically pick phases, associate arrivals, locate earthquakes, compute magnitudes, and generate bulletins. It also provides powerful tools for analyst review, including **dbpick**, which allows rapid scanning of continuous waveform data to identify and refine events. One of Antelope’s strengths is that archived data can be processed using the same workflows as real-time data.

Despite its technical strengths, Antelope did not become universal largely because it was commercial software. As a result, many observatories adopted hybrid workflows, often using Earthworm for real-time detection and SEISAN for manual processing and catalog refinement. This combination was used at many volcano observatories, including the **Montserrat Volcano Observatory** beginning in March 2001, when I was serving as Seismic Network Manager.
