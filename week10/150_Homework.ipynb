### 2. Build Your Processing Workflow

Create a **Jupyter notebook or Python script** that:

#### a. Load Data
- Load waveform data **one day at a time** (do not load everything at once).

#### b. Preprocess
- Identify and **clip large spikes** before computing RSAM  
  *(you may choose your clipping method, but justify it in comments)*

- If the dataset contains **more than 10 SEED IDs**:
  - Select only **vertical channels (Z components)**

---

#### c. Compute RSAM
- Compute RSAM for each trace using a fixed sampling interval (e.g., 60 s unless otherwise specified)

---

#### d. Save Outputs
- Save RSAM data to disk (CSV or pickle files)
- After processing, **compress all RSAM output files into a `.zip` archive**

---

### 3. Visualization

  - Multiple stations (if present) are clearly displayed
