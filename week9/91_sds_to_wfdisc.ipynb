This notebook:
1) selects MiniSEED files from an SDS archive using wildcards
2) previews matches
3) ingests them into an Antelope DB via `miniseed2db`

The heavy lifting is in `sds_antelope.py`.

```python
import os
from pathlib import Path
from obspy import UTCDateTime
from sds_antelope import SDSQuery, iter_sds_files, run_miniseed2db
os.system('source /opt/antelope/5.15/setup.sh')

SDS_ROOT = Path("/Volumes/classdata/SDS_Alaska")
HOME = os.path.expanduser('~')
WORKDIR = os.path.join(HOME, 'work', 'antelope_demo')
os.makedirs(WORKDIR, exist_ok=True)
DBNAME = os.path.join(WORKDIR, "dbredoubt2009")

query = SDSQuery(net="AV", sta="REF", cha="*", dtype="D")

start = UTCDateTime("2009-03-20T00:00:00")
end   = UTCDateTime("2009-03-23T23:59:59")

files = iter_sds_files(SDS_ROOT, query=query, starttime=start, endtime=end)

# preview
files_list = list(files)
print(f'Found {len(files_list)} miniseed files matching criteria')
for i in range(len(files_list)):
    print(f'{i}, {files_list[i]}')

# ingest
run_miniseed2db(files_list, DBNAME, dry_run=True, batch_size=25)
```

```python
# ingest
run_miniseed2db(files_list, DBNAME, dry_run=False, batch_size=25)
```

```python
query = SDSQuery(net="AV", sta="RSO", cha="*", dtype="D")

files = list(iter_sds_files(SDS_ROOT, query=query, starttime=start, endtime=end))

run_miniseed2db(files, DBNAME, dry_run=False, batch_size=25)
```
