
# Seismic Event Processing Workflow  
## Detection → Picking → Location → Magnitude

This notebook provides a comprehensive overview of seismic data processing workflows using:

- Antelope (Datascope)
- SEISAN
- ObsPy / Python
- NonLinLoc
- Pyrocko / Snuffler

---

# Key Perspective

**Antelope is arguably the most comprehensive and integrated seismic processing system**, especially for:

- Real-time monitoring
- Database-driven workflows
- Large network operations

However, it is:

- More complex to configure
- Heavily dependent on parameter files and database structure

Other tools (SEISAN, ObsPy, NonLinLoc, Pyrocko) are often simpler or more modular, and are widely used in research workflows.



# 1. Overall Workflow

```
Continuous Data → Detection → Picking → Location → Magnitude
```

| Step | Purpose | Output |
|------|--------|--------|
| Detection | Identify candidate events | Event times |
| Picking | Identify phase arrivals | Arrival times |
| Location | Estimate hypocenter | Origin |
| Magnitude | Estimate event size | Magnitude |



# 2. Detection

## Antelope
- `dbdetect`
- Fully integrated with Datascope database
- Scales well to continuous, real-time systems

## Python / ObsPy
- STA/LTA
- Template matching
- Machine learning
- Custom workflows (e.g., flovopy)

## Output
- Antelope: `detection` table
- Python: list of event times



# 3. Event Definition (Windowing)

After detection, waveform segments must be extracted.

## Tools
- Antelope: implicit via database
- SEISAN: `wavetool`, `mulplt`
- Python: ObsPy slicing

## Output
- Event waveform files
- Metadata structures (S-files, DataFrames, etc.)



# 4. Manual Picking

## Antelope
- `dbpick`
- Writes to `arrival` table
- Fully integrated with database

## SEISAN
- `eev`, `mulplt`
- Picks stored in S-files

## Pyrocko Snuffler
- Modern GUI picker
- Fast and intuitive
- Exportable picks

## Notes
- Picking is often the most time-consuming step
- Good visualization tools significantly improve efficiency



# 5. Event Location

## 5.1 Antelope: dbgrassoc

### Requirements
- `detection` / `arrival`
- `site` table
- Travel-time grid (`ttgrid`)
- Velocity model (used in grid)

### Workflow
```
dbdetect mydb
ttgrid mydb
dbgrassoc mydb
```

### Key Feature
- Uses precomputed travel-time grids
- Well-suited for real-time systems



## 5.2 SEISAN: hyp

### Minimum Requirement
- `STATION0.HYP`

Contains:
- Station coordinates
- 1-D velocity model
- Control parameters

### Optional Files
- `RESETUP`
- Station corrections

### Key Feature
- Travel times computed on-the-fly
- Simple and robust for offline work



## 5.3 NonLinLoc

### Role
Replaces traditional location step

```
Picks → NonLinLoc → Probabilistic location
```

### Advantages
- Handles complex velocity models
- Provides uncertainty estimates
- Widely used in volcano seismology



# 6. Magnitude Determination

## Antelope
- `dbevproc`

## SEISAN
- `hyp`, `eev`

## Types
- ML
- Mb
- Ms
- Coda magnitude

## Requirements
- Amplitude measurements
- Station calibration



# 7. Comparison of Systems

| Feature | Antelope | SEISAN | NonLinLoc | Pyrocko |
|--------|--------|--------|-----------|--------|
| Detection | ✔ | Limited | ✘ | ✘ |
| Picking | ✔ | ✔ | ✘ | ✔ |
| Location | ✔ | ✔ | ✔ | ✘ |
| Magnitude | ✔ | ✔ | ✘ | ✘ |
| Real-time | ✔ | ✘ | ✘ | ✘ |



# 8. Recommended Workflows

## Fully Antelope Workflow
```
Continuous → dbdetect → dbgrassoc → dbevproc
```

## SEISAN Workflow
```
Continuous → event windows → eev → hyp
```

## Modern Hybrid Workflow
```
Continuous (SDS)
    ↓
Detection (ObsPy / dbdetect)
    ↓
Picking (Snuffler / SEISAN / dbpick)
    ↓
Location (NonLinLoc or hyp)
    ↓
Magnitude
```



# 9. Key Insights

- Antelope = most comprehensive but complex
- SEISAN = simple and effective for offline work
- NonLinLoc = best for high-quality locations
- Snuffler = best interactive picker



# 10. Practical Advice

- Always verify station metadata
- Start with simple velocity models
- Use automated detection to reduce workload
- Keep workflows reproducible
- Separate storage (SDS) from processing
