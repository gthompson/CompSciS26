## Step 2 — Create a Response object for a USF Nanometrics station

````python
# TO DO: code for Response object for USF Nanometrics station from NRL

from obspy.core import Trace, Stream
from obspy.core.utcdatetime import UTCDateTime
from obspy.core.inventory import Inventory, Network, Station, Channel, Response
from obspy.clients.nrl import NRL
# from obspy.clients.fdsn import Client

# client = Client("IRIS")
nrl = NRL()

sensor_keys = ["Nanometrics", "Trillium Compact 120 (Vault, Posthole, OBS)", "754 V/m/s"]
datalogger_keys = ["Nanometrics", "Centaur", "40 Vpp (1)", "Off", "Linear phase", "100"]

resp_combined = nrl.get_response(datalogger_keys=datalogger_keys, sensor_keys=sensor_keys)
resp_sensor, _ = nrl._get_response("sensors", keys=sensor_keys)
resp_digitizer, _ = nrl._get_response("dataloggers", keys=datalogger_keys)
print("Combined response:", resp_combined)
````

## Step 3 - Create an ObsPy Inventory

The Inventory object should contain a list of 1 Network object.

The Network object should have a network code, description, and start_date, and contain a list of 7 Station objects.

Each Station object should have a station code, description, and coordinates, and a list of 3 Channel objects.

Each Channel object should have a channel code, a location code, coordinates (StationXML repeats them), depth, and sampling rate, and a Response object.


````python
# TO DO: Here is some starter code for the Inventory object. You will need to update it with station metadata from the Excel file and Response metadata from NRL.

from obspy.core.inventory import Inventory, Network, Station, Channel, Site
from obspy import UTCDateTime

# -----------------------------
# Network-level metadata
# -----------------------------
NETWORK_CODE = "1R"                     
NETWORK_DESCRIPTION = "KSC Seismic Network"
START_DATE = UTCDateTime(2026, 1, 7)    # TODO: update - this is not correct. it should be the date of the first station installation

# -----------------------------
# Channel metadata
# -----------------------------
CHANNEL_CODES = ["DHZ", "DHN", "DHE"]   # Google "SEED channel naming"
LOC_CODE = ""                           # usually blank unless multiple sensors
FS = 500.0                              # sample rate (Hz)
ELEV_M = 0.0                            # good enough - KSC is close to sea level
DEPTH_M = 0.75                          # burial depth (m) - holes were 2-3 feet deep
SENSOR = "Nanometrics Trillium C-PH 120"             
DIGITIZER = "Nanometrics Centaur"  

# Create Response object for USF Nanometrics station from NRL
nrl_usf = NRL()

usf_sensor_keys = ["Nanometrics", "Trillium Compact 120 (Vault, Posthole, OBS)", "754 V/m/s"]
usf_datalogger_keys = ["Nanometrics", "Centaur", "40 Vpp (1)", "Off", "Linear phase", "100"]

resp_usf = nrl_usf.get_response(datalogger_keys=usf_datalogger_keys, sensor_keys=usf_sensor_keys)
resp_usf_sensor, _ = nrl_usf._get_response("sensors", keys=usf_sensor_keys)
resp_usf_digitizer, _ = nrl_usf._get_response("dataloggers", keys=usf_datalogger_keys)


# Loop over rows in DataFrame - one row per station ("site name" is the station code)
stations = []
for index, row in df.iterrows():
    sta_code = row["Site Name"]
    sta_lat = row["lat"]
    sta_lon = row["lon"]
    sta_elev = ELEV_M
    sta_start_date = START_DATE
    site = Site(name=sta_code) # latitude=sta_lat, longitude=sta_lon, elevation=sta_elev)

    usf_station = Station(
        code=sta_code,
        latitude=sta_lat,
        longitude=sta_lon,
        elevation=sta_elev,
        creation_date=sta_start_date,
        site=site,
    )
    
    # Loop over CHANNEL_CODES and create a Channel object for each one.
    channels = []
    for chan_code in CHANNEL_CODES:
        if chan_code.endswith("Z"):
            azi, dip = 0.0, 90.0
        elif chan_code.endswith("N"):
            azi, dip = 0.0, 0.0
        elif chan_code.endswith("E"):
            azi, dip = 90.0, 0.0
        else:
            raise ValueError(f"Unexpected channel code: {chan_code}")
        
        channel = Channel(
            code=chan_code,
            location_code=LOC_CODE,
            sample_rate=FS,
            latitude=sta_lat,
            longitude=sta_lon,
            elevation=sta_elev,
            depth=DEPTH_M,
            azimuth=azi,
            dip=dip,
            response=resp_usf  # TODO: update - this is not correct. you will need to create a Response object for each station using NRL and use it here.
        )

        # append to channels list
        channels.append(channel)

    usf_station.channels = channels
    stations.append(usf_station)

network = Network(code=NETWORK_CODE, description=NETWORK_DESCRIPTION, start_date=START_DATE, stations=stations)
print(network)
````

## Step 4 — Build Inventory, plot, and write StationXML

````python
# TO DO

from obspy.core.inventory import Inventory

inv = Inventory(networks=[network], source="KSC Seismic Network")
print(inv)
inv.plot(projection="local", level="channel");
inv.write("usf_ksc_inventory.xml", format="STATIONXML")


# fill in - create Inventory object and add Network object to it
````

````python
# Access the network (assuming it's the first one in the inventory)
network = inv.networks[0]

# Iterate through every station in this network
for station in network.stations:
    # Iterate through every channel in this station
    for chan in station.channels:
        print(f"--- Station: {station.code} | Channel: {chan.code} ---")
        
        if chan.response:
            sensitivity = chan.response.instrument_sensitivity.value
            num_stages = len(chan.response.response_stages)
            print(f"Total Sensitivity: {sensitivity:.2f} counts/(m/s)")
            print(f"Number of Stages: {num_stages}")
        else:
            print("Response: No response attached!")
````
